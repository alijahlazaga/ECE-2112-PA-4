```python
import pandas as pd #Imports the pandas library and shortens pandas into pd
```

```python
import matplotlib.pyplot as plt #Imports the Matplotlib plotting tool and shortens matplotlib.pylot into plt
```

```python
df = pd.read_excel('board2.xlsx') #Reads excel file (.xlsx)
df
```

```python
df['Average'] = (df['Math'] + df['GEAS'] + df['Electronics'] + df['Communication'
]) / 4 #Produces a column that gets the sum of columns Math, GEAS, Electronics, and Communication scores and divides it by 4.
df
```

```python
#A. VISAYAS COMMUNICATION DATAFRAME
```

```python
VisComm = df.loc[(df['Hometown'] == 'Visayas' ) & (df['Track'] == 'Communication' )][['Name', 'Gender', 'Math', 'Electronics', 'Average']] #Locates students whose hometown is Visayas and in the Communication track. It names the new dataframe as VisComm
VisComm
```

```python
len(VisComm)
```

```python
#B. VISAYAS FEMALE DATAFRAME
```

```python
VisFemale = df.loc[(df['Hometown'] == 'Visayas') & (df['Gender'] == 'Female'), ['Name', 'Track', 'GEAS', 'Electronics', 'Average']] #Locates only students who are from visayas and are female from the dataframe. It then names the new dataframe as VisFemale
VisFemale
```

```python
VisFemale[VisFemale['Average'] >= 60] #Extract female students who got an average above or equal to  60
```

```python
# CATEGORY-AVERAGE VISUALIZATION
```

```python
#a. and b.
```

```python
TrackMean = df.groupby('Track')['Average'].mean() #Gets the mean of Average for the category, track.
TrackMean
```

```python
GenderMean = df.groupby('Gender')['Average'].mean() #Gets the mean of Average for the category, gender.
GenderMean
```

```python
HometownMean = df.groupby('Hometown')['Average'].mean() #Gets the mean of Average for the category, hometown.
HometownMean
```

```python
#c.
```

```python
#TrackMean
plt.figure(figsize = (20, 4)) #Creates an empty chart 20 inches wide, and 4 inches in height.
plt.subplot(1, 3, 1) #Creates a chart with 1 row, 3 columns, and selects position 1 of the chart for the codes to follow
plt.bar(TrackMean.index, TrackMean.values) #Draw bars by the horizontal axis and the mean average values needed.
plt.title('Mean Average by Track') #Creates a title for the chart
plt.xlabel('Track') #Labels the x-axis of the chart
plt.ylabel('Average') #Labels the y-axis of the chart
plt.ylim(0, 100) #Limits the y-axis from 0 to 100 only
```

```python
#GenderMean
plt.figure(figsize = (20, 4))#Creates an empty chart 20 inches wide, and 4 inches in height.
plt.subplot(1, 3, 2) #Creates a chart with 1 row, 3 columns, and selects position 2 of the chart for the codes to follow
plt.bar(GenderMean.index, GenderMean.values)#Draw bars by the horizontal axis and the mean average values needed.
plt.title('Mean Average by Gender')#Creates a title for the chart
plt.xlabel('Gender')#Labels the x-axis of the chart
plt.ylabel('Average')#Labels the y-axis of the chart
plt.ylim(0, 100)#Limits the y-axis from 0 to 100 only
```

```python
#HometownMean
plt.figure(figsize= (20,4))#Creates an empty chart 20 inches wide, and 4 inches in height.
plt.subplot(1, 3, 3) #Creates a chart with 1 row, 3 columns, and selects position 3 of the chart for the codes to follow
plt.bar(HometownMean.index, HometownMean.values)#Draw bars by the horizontal axis and the mean average values needed.
plt.title('Mean Average by Hometown')#Creates a title for the chart
plt.xlabel('Hometown')#Labels the x-axis of the chart
plt.ylabel('Average') #Labels the y-axis of the chart
plt.ylim(0, 100)#Limits the y-axis from 0 to 100 only
```
